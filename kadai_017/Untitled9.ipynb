```python


class Human:
    def __init__(self, age):
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print("大人である")
        else:
            print("大人でない")


h1 = Human(19)
h2 = Human(20)
h3 = Human(70)

array = [h1, h2, h3]

for i in range(0, 3):
    array[i].check_adult()
```
